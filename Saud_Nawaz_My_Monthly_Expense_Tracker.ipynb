```python
import pandas as pd

# expenses for this month
data = {
    'Category': ['Food', 'Transport', 'Books', 'Internet', 'Clothing'],

    'Amount_Spent': [3500, 1200, 1800, 1000, 2000]
}

df = pd.DataFrame(data)

income = 15000

print("========================================")
print("Monthly Expense Tracker")
print("Made by Saud Nawaz")
print("========================================")

print("\nExpenses:\n")

print(df.to_string(index=False))

total_spent = df['Amount_Spent'].sum()

money_left = income - total_spent

highest = df.loc[df['Amount_Spent'].idxmax()]

lowest = df.loc[df['Amount_Spent'].idxmin()]

print("\n----------------------------------------")
print("Summary")
print("----------------------------------------")

print("Income :", income)
print("Spent  :", total_spent)
print("Left   :", money_left)

print("\nMost money spent on:", highest['Category'])
print("Least money spent on:", lowest['Category'])

print("\nAdvice:")

if money_left < 0:
    print("You spent more than your income")
    print("Need to control spending")

elif money_left < 3000:
    print("Very little savings left")

else:
    print("Budget managed well this month")

print("\nProject Finished")
```
